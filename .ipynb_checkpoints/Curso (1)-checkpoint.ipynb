```python
#Lista de tarefas
tarefas = []
interface = "- Lista de tarefas -\n" \
"1 - Adicionar tarefa\n " \
"2 - Ver tarefas\n " \
"3 - Remover tarefa\n " \
"4 - Sair.\n"

def adicionar():
    tarefa = input("Digite a tarefa que será adicionada: ")
    tarefas.append(tarefa)
    return ("Tarefa adicionada!\n")
def listar():
    if len(tarefas) == 0:
        print("Nenhuma tarefa adicionada até o momento")
    else:
        for i in range (len(tarefas)):
            print(f"{i+1} - Tarefa = {tarefas[i]} ")
    print("O que deseja fazer agora?\n")
def remover():
    tarefa = input("Digite qual tarefa será removida (ou se quiser remover a ultima tarefa da lista digite 1): ")
    if tarefa == "1":
        tarefas.pop()
    else:
        tarefas.remove(tarefa)
    return ("Tarefa removida com sucesso!")
while True:
    print(interface)
    opcao = int(input("Escolha uma opção: "))
    if opcao == 1:
        adicionar()
    elif opcao == 2:
        listar()
    elif opcao == 3:
        remover()
    elif opcao == 4:
        print("Fechando lista....")
        break
    else:
        print("Opção digitada não existe")
```

```python
#Lançamento de dado
import random
contador_1 = 0
contador_2 = 0
contador_3 = 0
contador_4 = 0
contador_5 = 0
contador_6 = 0
for i in range(100):
    numero = random.randint(1,6) 
    if numero == 1:
        contador_1 +=1 
    elif numero == 2:
        contador_2 +=1
    elif numero == 3:
        contador_3 +=1
    elif numero == 4:
        contador_4 +=1
    elif numero == 5:
        contador_5 +=1
    else:
        contador_6 +=1
print(f"A quantidade de vezes que o número 1 se repetiu foi de {contador_1} vezes")
print(f"A quantidade de vezes que o número 2 se repetiu foi de {contador_2} vezes")
print(f"A quantidade de vezes que o número 3 se repetiu foi de {contador_3} vezes")
print(f"A quantidade de vezes que o número 4 se repetiu foi de {contador_4} vezes")
print(f"A quantidade de vezes que o número 5 se repetiu foi de {contador_5} vezes")
print(f"A quantidade de vezes que o número 6 se repetiu foi de {contador_6} vezes")
```

```python
#Sistema escolar 1.0
alunos = []
notas = []


menu = "- Sistema do Colégio Notre Dame -\n" \
"1 - Cadastrar aluno\n" \
"2 - Mostrar alunos\n" \
"3 - Calcular média\n" \
"4 - Pesquisar aluno\n" \
"5 - Alterar nota\n" \
"6 - Remover aluno\n" \
"7 - Sair\n"

def ler_nota():
    while True:
        nota = float(input("Digite a nota: "))

        if 0 <= nota <= 10:
            return nota

        print("Nota inválida.")

def cadastrar_alunos():
    aluno = input("Digite o nome do aluno: ")
    nota = ler_nota()
    alunos.append(aluno)
    notas.append(nota)
    print("Cadastro do aluno feito com sucesso!\n")
def mostrar_alunos(): 
    if len(alunos) == 0:
        print("Não existem alunos cadastrados.\n")
    else:
        print("\n===== ALUNOS =====")
        for i in range(len(alunos)):
            print(f"{i+1}. {alunos[i]} - Nota: {notas[i]}\n")

def calcular_media():
    media = 0
    if len(notas) == 0:
        print("Não existe nenhum aluno cadastrado.\n")
    else:
        for i in range(len(notas)):
            media += notas[i]
        media = media / len(notas)
        print(f"A média aritmétrica dos alunos cadastrados até o momento é {media:.2f}\n")

def pesquisar_aluno():
    if len(alunos) == 0:
        print("Não tem aluno cadastrados ainda.\n")
    else:
        
        pesquisar = input("Nome do aluno: ")
        if pesquisar in alunos:
            indice = alunos.index(pesquisar)

            print(f"Aluno: {alunos[indice]}")
            print(f"Nota: {notas[indice]}")

        else:
            print("Aluno não encontrado.\n")
def editar_nota():
    if len(alunos) == 0:
        print("Não tem alunos cadastrados ainda.")
    else:
        pesquisar = input("Digite o nome do aluno que deseja alterar a nota: ")
        if pesquisar in alunos:
            indice = alunos.index(pesquisar)
            print(f"A nota atual do aluno é {notas[indice]}")
            
            nota = ler_nota()
            notas[indice] = nota
            print("A nota foi alterada com sucesso!\n")
        else:
            print("Aluno inexistente.")
            
def remover_aluno():
    if len(alunos) == 0:
        print("Não existem alunos cadastrados ainda.")
    pesquisar = input("Digite o nome do aluno: ")
    if pesquisar in alunos:
        indice = alunos.index(pesquisar)
        print(f"O aluno {alunos[indice]} com nota {notas[indice]} será removido.")
        
        del alunos[indice]
        del notas[indice]
        
        print("Aluno removido com sucesso!")
    else:
        print("Aluno inexistente")
    
while True:
    print(menu)
    opcao = int(input("Digite o que deseja fazer: "))
    if opcao ==  1:
        while True:
            cadastrar_alunos()

            continuar = input("Cadastrar outro? (s/n): ")

            if continuar.lower() != "s":
                break
    elif opcao == 2:
        mostrar_alunos()
    elif opcao == 3:
        calcular_media()
    elif opcao == 4:
        pesquisar_aluno()
    elif opcao == 5:
        editar_nota()
    elif opcao == 6:
        remover_aluno()
    elif opcao == 7:
        break
    else:
        print("Opção não existe, digite novamente.\n")
```

```python
#Entendendo as dict, refatorando o sistema escolar 2.0
alunos = []

menu = "- Sistema do Colégio Notre Dame -\n" \
"1 - Cadastrar aluno\n" \
"2 - Mostrar alunos\n" \
"3 - Calcular média\n" \
"4 - Pesquisar aluno\n" \
"5 - Alterar nota\n" \
"6 - Remover aluno\n" \
"7 - Mostrar Status dos alunos\n" \
"8 - Sair"

def calcular_status(nota):
    if nota >= 7:
        return "Aprovado"
    else:
        return "Reprovado"

def localizar_aluno(nome):
    encontrou = False
    for aluno in alunos:
        if nome.lower() == aluno["nome"].lower():
            encontrou = True
            return aluno

        return None
def ler_nota():
    while True:
        try:
            nota = float(input("Digite a nota do aluno: "))
            if nota < 0 or nota > 10:
                print("Nota inválida(é menor que 0 ou maior que 10).")
            else:
                return nota
        except ValueError:
            print("Digite um número válido.")
def cadastrar_aluno():
    while True:
        nome = input("Nome: ").strip()
        if nome != "":
            break
        else:
            print("Nome inválido.")
    if localizar_aluno(nome) is not None:
        print("Já existe um aluno com esse nome.")
        return

    aluno = {
                "nome": nome,
                "nota": ler_nota(),
                "status": calcular_status("nota") 
            }
    alunos.append(aluno)

def mostrar_aluno():
    if not alunos:
        print("Não existem alunos cadastrados.")
    else:
        for i, aluno in enumerate(alunos, start=1):
            print(f"{i}. {aluno['nome']} - {aluno['nota']}")

def pesquisar_aluno():
    pesquisar = input("Digite o nome do aluno que deseja procurar: ")
    encontrou = False
    for aluno in alunos:
        if pesquisar.lower() == aluno["nome"].lower():
            print(f'Aluno : {aluno["nome"]}')
            print(f'Nota: {aluno["nota"]}')
            encontrou = True
    if not encontrou:
        print("Aluno não encontrado")

def alterar_nota():
    if not alunos:
        print("Não existem alunos cadastrados.")
        return
    
    pesquisar = input("Nome do aluno: ")
    encontrou = False
    for aluno in alunos:
        if pesquisar.lower() == aluno["nome"].lower():
            print(f"Aluno encontrado, Nota atual é : {aluno['nota']}\n ")
            encontrou = True
            nota = ler_nota()
            aluno["nota"] = nota
            print("Nota alterada com sucesso.")
            return
    if not encontrou:
        print("Aluno não encontrado.")
def remover_aluno():
    if not alunos:
        print("Não existem alunos cadastrados.")
        return
    nome = input("Digite o nome do aluno que deseja procurar: ").lower()
    aluno = localizar_aluno(nome)

    print(f"O aluno {aluno['nome']} com nota {aluno['nota']} será removido.")
    opcao = input("Deseja realmente remover esse aluno? (s/n)\n")
    if opcao.lower() == "s":
            alunos.remove(aluno)
            print("Aluno removido com sucesso.\n")
            return
    else:
            print("Remoção foi cancelada.\n")
            return

def calcular_media():
    if not alunos:
        print("Não existem alunos cadastrados")
        return
    soma = 0
    for aluno in alunos:
       soma += aluno["nota"]
    media = soma/len(alunos)
    print(f"A média dos alunos cadastrados até o momento é de {media:.2f}")
def mostrar_status():
    if not alunos:
        print("Não existem alunos cadastrados.")
        return
        
    for aluno in alunos:
        if aluno["nota"] >= 7:
            status = "Aprovado"
        else:
            status = "Reprovado"
   
    for aluno in alunos:
        print(f'{aluno["nome"]} - {aluno["nota"]} - {aluno["status"]}')
while True:
    print(menu)
    try:
        opcao = int(input("Digite o que deseja fazer:\n"))
        if opcao ==  1:
            while True:
                cadastrar_aluno()

                continuar = input("Cadastrar outro? (s/n): ")

                if continuar.lower() != "s":
                    break
        elif opcao == 2:
            mostrar_aluno()
        elif opcao == 3:
            calcular_media()
        elif opcao == 4:
            pesquisar_aluno()
        elif opcao == 5:
            alterar_nota()
        elif opcao == 6:
            remover_aluno()
        elif opcao == 7:
            mostrar_status()
        elif opcao == 8:
            break
        else:
            print("Opção não existe, digite novamente.\n")
    except ValueError:
        print("Valor digitado foi inválido.\n")


```

```python
#Atividade 16
import math

a = float(input("Digite o valor de a: "))

if a == 0:
    print("Erro: O valor de 'a' não pode ser zero. Isso não é uma equação do 2º grau.")
else:
    b = float(input("Digite o valor de b: "))
    c = float(input("Digite o valor de c: "))
    
    delta = (b ** 2) - (4 * a * c)
    print(f"\nO valor de Delta (Δ) é: {delta:.2f}")
    
    if delta < 0:
        print("A equação não possui raízes reais (Delta é negativo).")
        
    elif delta == 0:
        raiz = -b / (2 * a)
        print(f"A equação possui apenas uma raiz real: x = {raiz:.2f}")
        
    else:
        raiz1 = (-b + math.sqrt(delta)) / (2 * a)
        raiz2 = (-b - math.sqrt(delta)) / (2 * a)
        print("A equação possui duas raízes reais:")
        print(f"x1 = {raiz1:.2f}")
        print(f"x2 = {raiz2:.2f}")
```

```python
#Atividade 18
data = input("Digite uma data no formato dd/mm/aaaa: ")

partes = data.split("/")

if len(partes) == 3:
    dia = int(partes[0])
    mes = int(partes[1])
    ano = int(partes[2])

    bissexto = False

    if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
        bissexto = True

    if mes == 1 or mes == 3 or mes == 5 or mes == 7 or mes == 8 or mes == 10 or mes == 12:
        dias_maximos = 31
    elif mes == 4 or mes == 6 or mes == 9 or mes == 11:
        dias_maximos = 30
    elif mes == 2:
        if bissexto:
            dias_maximos = 29
        else:
            dias_maximos = 28
    else:
        dias_maximos = 0

    if (1 <= dia <= dias_maximos) and (1 <= mes <= 12):
        print(f"A data informada {data} é válida.")
    else:
        print("A data informada é inválida.")
else:
    print("Formato de data inválido.")
```

```python
#Atividade 37
codigos = []
alturas = []
pesos = []

while True:
    codigo = int(input("Digite o código do cliente (0 para encerrar): "))

    if codigo == 0:
        break

    altura = float(input("Digite a altura: "))
    peso = float(input("Digite o peso: "))

    codigos.append(codigo)
    alturas.append(altura)
    pesos.append(peso)

if len(codigos) > 0:
    indice_alto = alturas.index(max(alturas))
    indice_baixo = alturas.index(min(alturas))

    indice_gordo = pesos.index(max(pesos))
    indice_magro = pesos.index(min(pesos))

    media_altura = sum(alturas) / len(alturas)
    media_peso = sum(pesos) / len(pesos)

    print("\nRESULTADOS")

    print("Mais alto:")
    print("Código:", codigos[indice_alto])
    print("Altura:", alturas[indice_alto])

    print("\nMais baixo:")
    print("Código:", codigos[indice_baixo])
    print("Altura:", alturas[indice_baixo])

    print("\nMais gordo:")
    print("Código:", codigos[indice_gordo])
    print("Peso:", pesos[indice_gordo])

    print("\nMais magro:")
    print("Código:", codigos[indice_magro])
    print("Peso:", pesos[indice_magro])

    print("\nMédia das alturas:", round(media_altura, 2))
    print("Média dos pesos:", round(media_peso, 2))

else:
    print("Nenhum cliente cadastrado.")


```
